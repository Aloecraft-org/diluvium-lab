# Hello, Diluvium

Diluvium is Lua 5.5 with some additions. This notebook is the shortest
path to having run some, and a map of where the rest is.

**Run a cell with Ctrl+Enter.** Shift+Enter runs it and moves on.

```lua
print("hello, world")
```

## Cells share one kernel

Every cell runs in the same Lua state, in order. A global set in one
cell is there in the next — which is the whole point of a notebook and
the thing that makes running cells out of order confusing.

Run this one twice.

```lua
counter = (counter or 0) + 1
print("this cell has run " .. counter .. " time(s)")
```

## A cell ending in an expression shows its value

No `print` needed. The value appears as `Out[n]`, and tables show their
contents rather than `table: 0x1f2e0`.

```lua
2 ^ 10
```

```lua
{ name = "ada", tags = { "maths", "engines" }, year = 1843 }
```

`print` is left exactly as Lua defines it — a notebook that quietly
redefined it would teach you something that stops being true the moment
you run the same code in a terminal. The `Out[n]` echo is the notebook's
own affordance and is fair game.

## Errors keep the runtime's own words

With the line number, and a traceback where there is one. Under it you
will sometimes find a plain-English hint — the hint is *additive*, never
a rewrite, so the message stays the one you can search for.

```lua
local t = nil
print(t.field)
```

Errors are catchable, which on a WebAssembly build is not a given:
`pcall` needs `setjmp`/`longjmp` lowered onto WASM exception handling.

```lua
print(pcall(function() error("caught") end))
print(select(2, pcall(error, { code = 42 })).code)
```

## Strings interpolate

`$"..."` is Diluvium's, not stock Lua's. Everything after `::` inside a
slot goes to `string.format`.

```lua
local who, count = "world", 3
print($"hello {who}, {count} times")
print($"pi to four places: {math.pi::%.4f}")
```
